## Pandas Data Structures

There are 2 main data structures in Pandas:
- **Series**: 1D homogenous (single typed) array
- **DataFrame**: 2D tabular structure with potentially heterogenous (different typed) columns

### 2.2. Series

It is basically a one-dimensional labeled array, representing a single column or row in a dataset.

```python
# Create an empty Series
import numpy as np

empty_array = np.array([])

empty_series = pd.Series(empty_array)
empty_series
```

```python
arr = np.array(['pen', 'pencil', 'bag'])
pd.Series(arr, name='study_essentials')
```

```python
# We can also load data from dictionary
data = {'a': 10, 'b': 20, 'c': 30}

s = pd.Series(data, name='cool Series')
s
```

```python
# Locate data using label
s.loc['a']
```

### 2.3. DataFrame

we can think of a DataFrame as a bunch of Series objects put together to share the same index


```python
# data of 5 employees (name, gender, title)
employees = {
    'name': ['Alice Chen', 'Ben Sok', 'Clara Lim', 'Daniel Wong', 'Emily Tran'],
    'gender': ['F', 'M', 'F', 'M', 'F'],
    'title': ['Data Analyst', 'Project Manager', 'Software Engineer', 'Data Scientist', 'UI/UX Designer']
}
```

```python
# Create a DataFrame called employees_df
employees_df = pd.DataFrame(employees)

employees_df
```

### 2.4. Row Manipulation

```python
# age for the 5 employees
employees_ages = [30, 28, 42, 24, 23]
```

```python
# Create a new column called 'age'
employees_df['age'] = employees_ages
```

```python
employees_df
```

### 2.5 Combining DataFrames

```python
# university and gpa data for the 5 employees
education_data = {
    'university': ['AUPP', 'PIU', 'ITC', 'RUPP', 'AUPP'],
    'gpa': [3.66, 3.12, 2.12, 3.84, 3.12]
}
```

```python
# Create a new DataFrame called 'education_df'
education_df = pd.DataFrame(education_data)

education_df
```

```python
# Combine 'education_df' to 'employees_df' (use .concat() on column axis)
employees_df = pd.concat([employees_df, education_df], axis=1)
```

```python
employees_df
```

### 2.9 Descriptive Statistics

```python
# Statistics on age column
print("Age Statistics:")
print(f"  Count: {employees_df['age'].count()}")
print(f"  Mean: {employees_df['age'].mean()}")
print(f"  Median: {employees_df['age'].median()}")
print(f"  Mode: {employees_df['age'].mode().values}")
print(f"  Std Dev: {employees_df['age'].std()}")
print(f"  Min: {employees_df['age'].min()}")
print(f"  Max: {employees_df['age'].max()}")
```

```python
# Count the amount of each job title (using .value_counts())
title_amnt = employees_df['title'].value_counts()   # using .value_counts(), it also sorted for us in descending order (most value first)

title_amnt
```

#### Sorting

You can also do sorting operations, either sort by index or by value

```python
# Sort the amount of job title
title_amnt.sort_values(ascending=False)
```

```python
age_gpa_corr = employees_df['gpa'].corr(employees_df['age'])

age_gpa_corr
```

```python
# This is for next sessions!
import matplotlib as plt
import seaborn as sns

sns.regplot(data=employees_df, x='age', y='gpa')
```

### 2.11. Grouping Datasets

```python
# Group Job Title by Gender (using .groupby())
title_by_gender = employees_df.groupby('gender')['title'].value_counts()
```

```python
title_by_gender
```

```python
# Find the average GPA from each school
    # Perform mean aggregation on education_info

education_info = employees_df.groupby('university')['gpa'].mean()

education_info
```

Normally, reading a file is the initial step of doing data exploration.
