### Import Libraries

```python
# Import NumPy and Pandas
import pandas as pd
```

```python
# Read the downloaded dataset
data = pd.read_csv(path + "/valentine_gifts_satisfaction.csv")

data.head()
```

### Data Explanation

| Column | Type | Description |
|--------|------|-------------|
| `id` | Integer | Unique identifier (1-670) |
| `receiver_gender` | String | Gender of gift receiver (Female, Male, Non-binary) |
| `receiver_age` | Integer | Age of receiver (18-60 years) |
| `relationship_type` | String | Type of relationship (Dating, Married, First Date, Engaged, Complicated) |
| `months_together` | Integer | Duration of relationship in months (0-357) |
| `gift_item` | String | Specific gift given (18 different items) |
| `gift_category` | String | Category of gift (Sweets, Jewelry, Luxury, Collectibles, Gaming, Sentimental) |
| `price` | Float | Price of gift in USD ($5-$500) |
| `flower_count` | Integer | Number of flowers included (0, 1, 3, 6, 12, 24, 50, 67) |
| `days_prepared` | Integer | Days spent preparing the gift (0-200) |
| `receiver_satisfaction` | Integer | Satisfaction rating (1-5 scale, 1=very unhappy, 5=very happy) |

### Understanding the data

- find the shape of `data`
- get key information from our `data` using `.info()`
- get statistical summary of the `data` using `.describe()` 
- drop `id` column

```python
print("Shape of data:", data.shape)
```

```python
data.info()
```

```python
print("Statistical Summary of data:")
data.describe()
```

```python
data = data.drop(columns='id')
```

1. What is the proportion of receiver's gender?

```python
data['receiver_gender'].value_counts()
```

2. How many unique gifts are there? What are they?

```python
n_unique_gifts = data['gift_item'].nunique()
print(f"There are {n_unique_gifts} unique gifts")

data['gift_item'].unique()
```

3. What gift category were gifted the most? Which one were gifted the least?

```python
# Use .value_counts() to show the occurences of each elements
data['gift_category'].value_counts()
```

```python
most_gifted = data['gift_category'].value_counts().idxmax() # .idxmax() returns the index of the maximum value
least_gifted = data['gift_category'].value_counts().idxmin()

print(f"Most Gifted: {most_gifted}\nMost Gifted: {least_gifted}")
```

4. What was the most popular gift item?

```python
data['gift_item'].value_counts()
```

5. What was the average price spent on Labubu Doll?

```python
data[data['gift_item'] == 'Labubu Doll']['price'].mean()
```

6. On average, what item is the most expensive gift?

```python
data.groupby('gift_item')['price'].mean().sort_values(ascending=False)
```

7. Find all gifts given to people aged 40 or older

```python
data[data['receiver_age'] >= 40]['gift_item'].value_counts()
```

8. How many people did not get their partner a flower?

```python
(data['flower_count'] == 0).sum()

# orrrr
# len(data[data['flower_count'] == 0])
```

9. Find all receivers who received a gift of less than $10 in price, and still fully satisfied with their gift.

```python
data[(data['price'] < 10) & (data['receiver_satisfaction'] == 5)]
# Note: don't forget to wrap each condition in a parenthesis ()
# Also, Pandas support bit-wise operations (& |)
```

10. Display all the data for the couple who have been together for over 300 months

```python
data[data['months_together'] > 300]
```

11. Calculate the average satisfaction for each gift item

```python
data.groupby('gift_item')['receiver_satisfaction'].mean()
```

12. (Hard) Find the couple who have been together the longest, and still not engaged or married yet
> Hint: 
> - Filter using `~` (NOT operator) and `.isin()`
> - Use `.nlargest(n, columns, keep(optional))` to find the first `n` rows with the largest values 
>    > - n: number of rows to return
>    > - columns: column name
>    > - keep (optional): whether to retrieve the `first`, `last`, or retrieve `all`, in case of duplicates

```python
data[~data['relationship_type'].isin(['Married', 'Engaged'])].nlargest(n=1, columns='months_together', keep='all')
```
